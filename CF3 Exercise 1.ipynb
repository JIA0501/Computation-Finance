Question 1.1

Let $V(t,S_t) := C_d(t,S_t)$. The binary call payoff is $V(T,S_T) = 1_{[K, \infty]} (S_T)$.
We apply Ito's formula to $V(t,S_t)$ and get

$$
dV= \frac{\partial V}{\partial t} \, dt + \frac{\partial V}{\partial s} \, dS_t + \frac{1}{2} \frac{\partial^2 V}{\partial s^2} \, d[S_t]$$

with 

\begin{align}
dS_t &= rS_t \, dt + \sigma S_t \, dB_t \\
d[S_t] &= \sigma^2 S_t^2 \, dt
\end{align}

Substituting gives

$$
dV = (\frac{\partial V}{\partial t}  + rS_t \frac{\partial V}{\partial s} + \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 V}{\partial s^2} ) dt + \sigma S_t \frac{\partial V}{\partial s} \, dB_t
$$

Now consider the self-financing portfolio
$$
\Pi_t = -V(t,S_t) + \Delta_t S_t
$$


We use the fact that the portfolio is self-financing so

$$
d\Pi_t = -dV + \Delta_t \, dS_t
$$

Substituting and rearranging gives 

$$
d\Pi_t = (- \frac{\partial V}{\partial t}  - rS_t \frac{\partial V}{\partial s} - \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 V}{\partial s^2} + rS_t\Delta_t ) \, dt + \sigma S_t (\Delta_t - \frac{\partial V}{\partial s}) \, dB_t
$$

The portfolio risk is entirely contained in the stochastic term. So we choose

$$
\Delta_t = \frac{\partial V}{\partial s}(t, S_t)
$$

to eliminate the $dB_t$ term.
Then the portfolio is riskless and is given as

$$
d\Pi_t =  (- \frac{\partial V}{\partial t}  - \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 V}{\partial s^2} ) \, dt
$$

Since it is now riskless it must earn the risk-free rate

$$
d\Pi_t = r \Pi_t \, dt = r(-V + S_t \frac{\partial V}{\partial s} )
$$

So

$$
- \frac{\partial V}{\partial t}  - \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 V}{\partial s^2} =  r(-V + S_t \frac{\partial V}{\partial s} )
$$

This then is equivalent to the Black-Scholes PDE

$$
\frac{\partial V}{\partial t}  + \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 V}{\partial s^2} + r S_t \frac{\partial V}{\partial s} - rV = 0
$$

Thus the Black-Scholes PDE satisfied by $C_d(t,s)$ is 

$$
\frac{\partial C_d}{\partial t}(t,s) + \frac{1}{2} \sigma^2 S_t^2 \frac{\partial^2 C_d}{\partial s^2}(t,s) + r S_t \frac{\partial C_d}{\partial s}(t,s) - rC_d(t,s) = 0
$$

with $0 \leq t < T, \, s>0$.

The terminal condition is

$$
C_d(t,s) = 1_{[K, \infty]}(s) =

\begin{cases}
1, & s \geq K, \\
0, & s < K.
\end{cases}
$$

Question 1.2

We have that the payoff of the binary call is

$$
h(S_T) = \mathbf{1}_{[K,\infty)}(S_T)
$$

so the risk-neutral pricing formula gives

$$
C_d(t,x) = V(t,x) = \mathbb{E}^{\mathbb{Q}}[ e^{-r(T-t)} h(S_T) \mid S_t = x ] = \mathbb{E}^{\mathbb{Q}}[ e^{-r(T-t)} \mathbf{1}_{[K,\infty)}(S_T) \mid S_t = x ]
$$

So

$$
C_d(t,x) = e^{-r(T-t)} \mathbb{Q}(S_T \ge K \mid S_t = x)
$$

We have the solution of the SDE as 

$$
S_T = x \exp( (r-\tfrac{1}{2}\sigma^2)(T-t) + \sigma(B_T-B_t) )
$$

Taking logarithms gives

$$
\log S_T = \log x + (r-\tfrac{1}{2}\sigma^2)(T-t) + \sigma(B_T-B_t)
$$

Since

$$
B_T - B_t \sim \mathcal{N}(0, T-t)
$$

it follows that

$$
\log S_T \sim \mathcal{N}( \log x + (r-\tfrac{1}{2}\sigma^2)(T-t), \, \sigma^2(T-t) )
$$

We now have

$$
\mathbb{Q}(S_T \ge K \mid S_t = x) = \mathbb{Q}(\log S_T \ge \log K \mid S_t = x)
$$


This gives

$$
\mathbb{Q}(S_T \geq K \mid S_t = x) = \mathbb{Q}(\frac{\log S_T - \log x - (r-\tfrac{1}{2}\sigma^2)(T-t)}{|\sigma|\sqrt{T-t}} \geq
\frac{\log K - \log x - (r-\tfrac{1}{2}\sigma^2)(T-t)}{|\sigma|\sqrt{T-t}})
$$


We then obtain using the CDF $\Phi$

$$
\mathbb{Q}(S_T \ge K \mid S_t = x) = \Phi(\frac{  (r-\tfrac{1}{2}\sigma^2)(T-t) + \log(x/K)  }{ |\sigma|\sqrt{T-t} })
$$


Thus the solution of the PDE is given as 

$$
C_d(t,x) = e^{-r(T-t)}\Phi(d_-(T-t))
$$

where 

$$
d_-(T-t) := \frac{  (r-\tfrac{1}{2}\sigma^2)(T-t) + \log(x/K) }{ |\sigma|\sqrt{T-t} }, \quad 0 \leq t < T
$$

Question 1.3

The Monte Carlo estimator for the option price is

$$
\hat{C}_d^{MC} = e^{-r(T-t)}\frac{1}{N}\sum_{i=1}^N \mathbf{1}_{\{S_T^{(i)} \geq K\}}
$$

The closed-form solution is
$$
C_d(t,x) = e^{-r(T-t)}\Phi(\frac{  (r-\tfrac{1}{2}\sigma^2)(T-t) + \log(x/K) }{ |\sigma|\sqrt{T-t} })
$$

We compute both at time $t=0$ using the following code

```python
import numpy as np
from scipy.stats import norm

# parameters
S0 = 1.0
K = 1.0
r = 0.05
sigma = 0.2
T = 1.0
N = 100000

# Monte Carlo
Z = np.random.randn(N)
ST = S0 * np.exp((r - 0.5 * sigma**2) * T + sigma * np.sqrt(T) * Z)
payoff = (ST >= K).astype(float)
price_mc = np.exp(-r * T) * np.mean(payoff)

# closed form
d_minus = (np.log(S0 / K) + (r - 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
price_cf = np.exp(-r * T) * norm.cdf(d_minus)

print("Monte Carlo price:", price_mc)
print("Closed-form price:", price_cf)
print("Absolute error:", abs(price_mc - price_cf))
```

```python
import matplotlib.pyplot as plt

# parameters
K = 1.0
r = 0.05
sigma = 0.2
T = 1.0

Smax = 3.0
M = 200
N = 200

S = np.linspace(0, Smax, M + 1)
t = np.linspace(0, T, N + 1)

dS = S[1] - S[0]
dt = T / N

def implicit_step(v_old, a, b, c, left_new, right_new):
    m = len(v_old)

    main = 1.0 - b
    lower = -a[1:]
    upper = -c[:-1]

    A = np.diag(main) + np.diag(lower, -1) + np.diag(upper, 1)

    rhs = v_old.copy()
    rhs[0] += a[0] * left_new
    rhs[-1] += c[-1] * right_new

    return np.linalg.solve(A, rhs)


def crank_nicolson_step(v_old, a, b, c, left_old, left_new, right_old, right_new):
    m = len(v_old)

    main_1 = 1.0 - 0.5 * b
    lower_1 = -0.5 * a[1:]
    upper_1 = -0.5 * c[:-1]

    main_2 = 1.0 + 0.5 * b
    lower_2 = 0.5 * a[1:]
    upper_2 = 0.5 * c[:-1]

    A1 = np.diag(main_1) + np.diag(lower_1, -1) + np.diag(upper_1, 1)
    A2 = np.diag(main_2) + np.diag(lower_2, -1) + np.diag(upper_2, 1)

    rhs = A2 @ v_old
    rhs[0] += 0.5 * a[0] * (left_old + left_new)
    rhs[-1] += 0.5 * c[-1] * (right_old + right_new)

    return np.linalg.solve(A1, rhs)


def solve_digital_option(K, r, sigma, T, Smax, M, N, scheme="implicit"):
    S = np.linspace(0, Smax, M + 1)
    t = np.linspace(0, T, N + 1)

    dt = T / N

    V = np.zeros((N + 1, M + 1))

    # terminal payoff
    V[-1, :] = (S >= K).astype(float)

    # boundary conditions
    for n in range(N + 1):
        tau = T - t[n]
        V[n, 0] = 0.0
        V[n, -1] = np.exp(-r * tau)

    j = np.arange(1, M)
    a = dt * (0.5 * sigma**2 * j**2 - 0.5 * r * j)
    b = dt * (-sigma**2 * j**2 - r)
    c = dt * (0.5 * sigma**2 * j**2 + 0.5 * r * j)

    for n in range(N - 1, -1, -1):
        v_old = V[n + 1, 1:M].copy()

        if scheme == "implicit":
            V[n, 1:M] = implicit_step(
                v_old,
                a,
                b,
                c,
                V[n, 0],
                V[n, -1]
            )

        elif scheme == "cn":
            V[n, 1:M] = crank_nicolson_step(
                v_old,
                a,
                b,
                c,
                V[n + 1, 0],
                V[n, 0],
                V[n + 1, -1],
                V[n, -1]
            )

    return S, t, V


# implicit scheme
S_imp, t_imp, V_imp = solve_digital_option(K, r, sigma, T, Smax, M, N, scheme="implicit")

# Crank-Nicolson scheme
S_cn, t_cn, V_cn = solve_digital_option(K, r, sigma, T, Smax, M, N, scheme="cn")

# price at S0
S0 = 1.0
price_imp = np.interp(S0, S_imp, V_imp[0, :])
price_cn = np.interp(S0, S_cn, V_cn[0, :])

print("Implicit price =", price_imp)
print("Crank-Nicolson price =", price_cn)

# sensitivity analysis
sigmas = np.linspace(0.1, 0.6, 10)
rates = np.linspace(0.0, 0.1, 10)
maturities = np.linspace(0.25, 2.0, 10)

prices_sigma_imp = []
prices_sigma_cn = []

for sig in sigmas:
    S_tmp, t_tmp, V_tmp = solve_digital_option(K, r, sig, T, Smax, M, N, scheme="implicit")
    prices_sigma_imp.append(np.interp(S0, S_tmp, V_tmp[0, :]))

    S_tmp, t_tmp, V_tmp = solve_digital_option(K, r, sig, T, Smax, M, N, scheme="cn")
    prices_sigma_cn.append(np.interp(S0, S_tmp, V_tmp[0, :]))

plt.figure(figsize=(8, 5))
plt.plot(sigmas, prices_sigma_imp, label="Implicit")
plt.plot(sigmas, prices_sigma_cn, label="Crank-Nicolson")
plt.xlabel("sigma")
plt.ylabel("Price")
plt.title("Sensitivity with respect to sigma")
plt.legend()
plt.grid(True)
plt.show()


prices_r_imp = []
prices_r_cn = []

for rr in rates:
    S_tmp, t_tmp, V_tmp = solve_digital_option(K, rr, sigma, T, Smax, M, N, scheme="implicit")
    prices_r_imp.append(np.interp(S0, S_tmp, V_tmp[0, :]))

    S_tmp, t_tmp, V_tmp = solve_digital_option(K, rr, sigma, T, Smax, M, N, scheme="cn")
    prices_r_cn.append(np.interp(S0, S_tmp, V_tmp[0, :]))

plt.figure(figsize=(8, 5))
plt.plot(rates, prices_r_imp, label="Implicit")
plt.plot(rates, prices_r_cn, label="Crank-Nicolson")
plt.xlabel("r")
plt.ylabel("Price")
plt.title("Sensitivity with respect to r")
plt.legend()
plt.grid(True)
plt.show()


prices_T_imp = []
prices_T_cn = []

for TT in maturities:
    S_tmp, t_tmp, V_tmp = solve_digital_option(K, r, sigma, TT, Smax, M, N, scheme="implicit")
    prices_T_imp.append(np.interp(S0, S_tmp, V_tmp[0, :]))

    S_tmp, t_tmp, V_tmp = solve_digital_option(K, r, sigma, TT, Smax, M, N, scheme="cn")
    prices_T_cn.append(np.interp(S0, S_tmp, V_tmp[0, :]))

plt.figure(figsize=(8, 5))
plt.plot(maturities, prices_T_imp, label="Implicit")
plt.plot(maturities, prices_T_cn, label="Crank-Nicolson")
plt.xlabel("T")
plt.ylabel("Price")
plt.title("Sensitivity with respect to T")
plt.legend()
plt.grid(True)
plt.show()


# surface c(S,t)
T_grid, S_grid = np.meshgrid(t_cn, S_cn)

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(S_grid, T_grid, V_cn.T)
ax.set_xlabel("S")
ax.set_ylabel("t")
ax.set_zlabel("c(S,t)")
ax.set_title("Digital option surface c(S,t)")
plt.show()


# delta as function of S
delta_imp = np.gradient(V_imp[0, :], S_imp)
delta_cn = np.gradient(V_cn[0, :], S_cn)

plt.figure(figsize=(8, 5))
plt.plot(S_imp, delta_imp, label="Implicit")
plt.plot(S_cn, delta_cn, label="Crank-Nicolson")
plt.xlabel("S")
plt.ylabel("Delta")
plt.title("Delta as a function of S")
plt.legend()
plt.grid(True)
plt.show()
```

Knock-out Barrier options

Question 1

We start from the following equation, when $K \leq B$, we have

$$
C(H) = e^{-r(T-t)} 1_{\{M_0^t < B\}} \mathbb{E}\Bigg[\Bigg(x \frac{S_{T-t}}{S_0} - K\Bigg)^+ 1_{\{x \max_{0 \leq u \leq T-t} \frac{S_u}{S_0} < B\}}\Bigg]_{x=S_t}
$$

Using that
$$
S_T = S_t \frac{S_{T-t}}{S_0}
$$
and
$$
S_t \max_{0 \leq u \leq T-t} \frac{S_u}{S_0} = \max_{t \leq u \leq T} S_u = M_t^T
$$

we obtain after conditioning on the filtration $\mathcal{F}_t$

$$
C(H) = e^{-r(T-t)} \mathbb{E}\Big[(S_T-K)^+ 1_{\{M_0^T < B\}} \mid \mathcal F_t\Big]
$$

Since the event $\{M_0^T < B\}$ implies $\{M_0^t < B\}$ and $M_0^t$ is $\mathcal F_t$-measurable, this gives

$$
C(H) = e^{-r(T-t)} 1_{\{M_0^t < B\}} \mathbb{E}\Big[(S_T-K)^+ 1_{\{M_t^T < B\}} \mid \mathcal F_t\Big]
$$

Under the risk-neutral measure $\mathbb{Q}$ we have


$$
S_T = S_t \exp\big((r-\sigma^2/2)(T-t) + \sigma(W_T-W_t)\big)
$$

Since $W_T-W_t$ is independent of the filtration $\mathcal{F}_t$, we get

$$
C(H) = e^{-r(T-t)} 1_{\{M_0^t < B\}} \mathbb{E}\Big[\big(S_t e^{(r-\sigma^2/2)(T-t)+\sigma W_{T-t}} - K\big)^+ 1_{\{\max_{0 \leq u \leq T-t} S_t e^{(r-\sigma^2/2)u+\sigma W_u} < B\}}\Big]
$$

Now the barrier condition is equivalent to

$$
\max_{0 \leq u \leq T-t} S_t e^{(r-\sigma^2/2)u+\sigma W_u} < B
$$

$$
\iff \max_{0 \leq u \leq T-t} \big((r-\sigma^2/2)u + \sigma W_u\big) < \log(B/S_t)
$$

To simplify the notation we rewrite it in the following way

$$
\big(S_t e^{(r-\sigma^2/2)(T-t)+\sigma W_{T-t}} - K\big)^+ = \big(S_t e^X - K\big)^+
$$

with

$$
X = (r-\sigma^2/2)(T-t) + \sigma W_{T-t}
$$

which is non-zero only if

$$
X > \log(K/S_t)
$$

So this gives

$$
C(H) = e^{-r(T-t)} 1_{\{M_0^t < B\}} \mathbb{E}\Big[\big(S_t e^X - K\big) 1_{\{X > \log(K/S_t)\}} 1_{\{\max_{0 \leq u \leq T-t} ((r-\sigma^2/2)u+\sigma W_u) < \log(B/S_t)\}}\Big]
$$

Since $K \leq B$, we have $\log(K/S_t) \leq \log(B/S_t)$, the indicators imply that $x \in [\log(K/S_t), \log(B/S_t)]$. Using the density of $X$ under the barrier condition, we write

$$
C(H)= e^{-r(T-t)} 1_{\{M_0^t < B\}} \int_{\log(K/S_t)}^{\log(B/S_t)} (S_t e^x - K)\, f(x) \, dx
$$

where

$$
f(x) = \frac{d}{dx} \mathbb{Q}\Big(X \leq x,\ \max_{0 \leq u \leq T-t}((r-\sigma^2/2)u+\sigma W_u) < \log(B/S_t)\Big)
$$

We first note that since

$$
X = (r-\sigma^2/2)(T-t) + \sigma W_{T-t}
$$

we have

$$
X \sim \mathcal{N}\big((r-\sigma^2/2)(T-t),\ \sigma^2 (T-t)\big)
$$

and therefore

$$
\frac{d}{dx} \mathbb{Q}(X \leq x)
=
\frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big)
$$

To incorporate the barrier condition, we write

$$
\mathbb{Q}\Big(X \leq x,\ \max_{0 \leq u \leq T-t}((r-\sigma^2/2)u+\sigma W_u) < \log(B/S_t)\Big)
$$

$$
=
\mathbb{Q}(X \leq x)
-
\mathbb{Q}\Big(X \leq x,\ \max_{0 \leq u \leq T-t}((r-\sigma^2/2)u+\sigma W_u) \geq \log(B/S_t)\Big)
$$

The second term represents paths that cross the barrier and is obtained via the reflection principle at the level $\log(B/S_t)$

$$
\frac{d}{dx} \mathbb{Q}\Big(X \leq x,\ \max_{0 \leq u \leq T-t}((r-\sigma^2/2)u+\sigma W_u) \geq \log(B/S_t)\Big)
$$

$$
=
e^{\frac{2(r-\sigma^2/2)\log(B/S_t)}{\sigma^2}}
\frac{1}{\sigma\sqrt{T-t}}
\phi\Big(\frac{x-2\log(B/S_t)-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big)
$$

Then we finally get the following expression for the density



$$
f(x) = \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) -e^{\frac{2(r-\sigma^2/2)\log(B/S_t)}{\sigma^2}}\frac{1}{\sigma\sqrt{T-t}}\phi\Big(\frac{x-2\log(B/S_t)-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big)
$$

Substituting this into the integral gives
$$
C(H) = e^{-r(T-t)} 1_{\{M_0^t < B\}} \int_{\log(K/S_t)}^{\log(B/S_t)} (S_t e^x - K)\Bigg[\frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) - e^{\frac{2(r-\sigma^2/2)\log(B/S_t)}{\sigma^2}} \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-2\log(B/S_t)-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big)\Bigg] \, dx
$$

We now split this into four terms

$$
C(H) = I_1 - I_2 - I_3 + I_4
$$

\begin{align}
I_1 &= e^{-r(T-t)} 1_{\{M_0^t < B\}} S_t \int_{\log(K/S_t)}^{\log(B/S_t)} e^x \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) \, dx \\
I_2 &= e^{-r(T-t)} 1_{\{M_0^t < B\}} S_t e^{\frac{2(r-\sigma^2/2)\log(B/S_t)}{\sigma^2}} \int_{\log(K/S_t)}^{\log(B/S_t)} e^x \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-2\log(B/S_t)-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) \, dx \\
I_3 &= e^{-r(T-t)} 1_{\{M_0^t < B\}} K \int_{\log(K/S_t)}^{\log(B/S_t)} \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) \, dx \\
I_4 &= e^{-r(T-t)} 1_{\{M_0^t < B\}} K e^{\frac{2(r-\sigma^2/2)\log(B/S_t)}{\sigma^2}} \int_{\log(K/S_t)}^{\log(B/S_t)} \frac{1}{\sigma\sqrt{T-t}} \phi\Big(\frac{x-2\log(B/S_t)-(r-\sigma^2/2)(T-t)}{\sigma\sqrt{T-t}}\Big) \, dx
\end{align}

For $I_1$ we complete the square and recognize the normal cdf and get

$$
I_1 = S_t 1_{\{M_0^t < B\}} \Big[\Phi(\delta_{T-t}^+(S_t/K)) - \Phi(\delta_{T-t}^+(S_t/B))\Big]
$$

For $I_3$ we directly get

$$
I_3 = e^{-r(T-t)} K 1_{\{M_0^t < B\}} \Big[\Phi(\delta_{T-t}^-(S_t/K)) - \Phi(\delta_{T-t}^-(S_t/B))\Big]
$$

For $I_2$ we use the substitution

$$
y = 2\log(B/S_t) - x
$$

which gives

$$
x = \log(K/S_t) \Rightarrow y = \log(B^2/(K S_t))
\qquad
x = \log(B/S_t) \Rightarrow y = \log(B/S_t)
$$

so after simplification

$$
I_2 = S_t 1_{\{M_0^t < B\}} \Big(\frac{B}{S_t}\Big)^{1+2r/\sigma^2} \Big[\Phi(\delta_{T-t}^+(B^2/(K S_t))) - \Phi(\delta_{T-t}^+(B/S_t))\Big]
$$

Similarly,

$$
I_4 = e^{-r(T-t)} K 1_{\{M_0^t < B\}} \Big(\frac{S_t}{B}\Big)^{1-2r/\sigma^2} \Big[\Phi(\delta_{T-t}^-(B^2/(K S_t))) - \Phi(\delta_{T-t}^-(B/S_t))\Big]
$$

Substituting these expressions into $C(H) = I_1 - I_2 - I_3 + I_4$ gives

$$
C(H) = S_t 1_{\{M_0^t < B\}} \Big\{\Phi(\delta_{T-t}^+(S_t/K)) - \Phi(\delta_{T-t}^+(S_t/B)) - \Big(\frac{B}{S_t}\Big)^{1+2r/\sigma^2}\Big[\Phi(\delta_{T-t}^+(B^2/(K S_t))) - \Phi(\delta_{T-t}^+(B/S_t))\Big]\Big\}
$$

$$
\qquad - e^{-r(T-t)} K 1_{\{M_0^t < B\}} \Big\{\Phi(\delta_{T-t}^-(S_t/K)) - \Phi(\delta_{T-t}^-(S_t/B)) - \Big(\frac{S_t}{B}\Big)^{1-2r/\sigma^2}\Big[\Phi(\delta_{T-t}^-(B^2/(K S_t))) - \Phi(\delta_{T-t}^-(B/S_t))\Big]\Big\}
$$

where

$$
\delta_\tau^\pm(z) = \frac{1}{\sigma\sqrt{\tau}}\big(\log z + (r \pm \sigma^2/2)\tau\big), \quad z > 0
$$


Question 2

We compare the analytical price $C(H)$ with a Monte Carlo estimator for the discretely monitored up-and-out call.

Let

$$
\Delta t = \frac{T-t}{m}
$$

and under $\mathbb{Q}$ simulate

$$
S_{t_{k+1}} = S_{t_k} \exp\big((r-\sigma^2/2)\Delta t + \sigma \sqrt{\Delta t} Z_{k+1}\big)
$$

with $Z_{k+1} \sim \mathcal{N}(0,1)$ i.i.d.

The Monte Carlo estimator is

$$
\hat C_m(H) = e^{-r(T-t)} \frac{1}{N} \sum_{i=1}^N (S_T^{(i)} - K)^+ 1_{\{\max_{0 \leq k \leq m} S_{t_k}^{(i)} < B\}}
$$

Since this is a discretely monitored barrier option, we use the correction 

$$
C_m(H) = C\big(H e^{\beta_1 \sigma \sqrt{T/m}}\big) + o(1/\sqrt{m})
$$

with $\beta_1 \approx 0.5826$. We define the shifted barrier

$$
B_m = B e^{\beta_1 \sigma \sqrt{T/m}}
$$

and compare $\hat C_m(H)$ with the analytical price $C(H)$ evaluated at $B_m$. We then vary $m$ and compute

$$
|\hat C_m(H) - C(H)|
\qquad
\text{and}
\qquad
|\hat C_m(H) - C(H)\big|_{B=B_m}|
$$

to study convergence. We now do the MC simulation and run a convergence analysis on the MC method.

```python


# Closed-form price 

def d_plus(z, T, r, sigma):
    return (np.log(z) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))

def d_minus(z, T, r, sigma):
    return (np.log(z) + (r - 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))

def barrier_call_closed(S, K, B, r, sigma, T):
    if S >= B:
        return 0.0

    part1 = S * (norm.cdf(d_plus(S / K, T, r, sigma)) - norm.cdf(d_plus(S / B, T, r, sigma)))

    part2 = S * (B / S)**(1 + 2 * r / sigma**2) * (
        norm.cdf(d_plus(B**2 / (K * S), T, r, sigma)) -
        norm.cdf(d_plus(B / S, T, r, sigma))
    )

    part3 = K * np.exp(-r * T) * (
        norm.cdf(d_minus(S / K, T, r, sigma)) -
        norm.cdf(d_minus(S / B, T, r, sigma))
    )

    part4 = K * np.exp(-r * T) * (S / B)**(1 - 2 * r / sigma**2) * (
        norm.cdf(d_minus(B**2 / (K * S), T, r, sigma)) -
        norm.cdf(d_minus(B / S, T, r, sigma))
    )

    return part1 - part2 - part3 + part4


# Monte Carlo price

def barrier_call_mc(S0, K, B, r, sigma, T, m, N, seed=1):
    np.random.seed(seed)
    dt = T / m
    payoffs = []

    for i in range(N):
        S = S0
        hit_barrier = False

        for k in range(m):
            Z = np.random.randn()
            S = S * np.exp((r - 0.5 * sigma**2) * dt + sigma * np.sqrt(dt) * Z)

            if S >= B:
                hit_barrier = True
                break

        if hit_barrier:
            payoffs.append(0.0)
        else:
            payoffs.append(np.exp(-r * T) * max(S - K, 0.0))

    payoffs = np.array(payoffs)
    price = np.mean(payoffs)
    se = np.std(payoffs, ddof=1) / np.sqrt(N)

    return price, se


# Parameters

S0 = 100
K = 95
B = 120
r = 0.03
sigma = 0.20
T = 1.0
beta1 = 0.5826


# comparison for fixed m

m = 252
N = 20000

C_exact = barrier_call_closed(S0, K, B, r, sigma, T)
C_mc, se_mc = barrier_call_mc(S0, K, B, r, sigma, T, m, N)

B_m = B * np.exp(beta1 * sigma * np.sqrt(T / m))
C_corr = barrier_call_closed(S0, K, B_m, r, sigma, T)

print("Continuous analytical price:", C_exact)
print("Monte Carlo price:", C_mc)
print("Monte Carlo standard error:", se_mc)
print("Shifted barrier B_m:", B_m)
print("Corrected analytical price:", C_corr)
print("Absolute error vs continuous price:", abs(C_mc - C_exact))
print("Absolute error vs corrected price:", abs(C_mc - C_corr))


# comparison in sigma

sigma_values = [0.10, 0.15, 0.20, 0.25, 0.30, 0.35]
closed_prices = []
mc_prices = []

for sig in sigma_values:
    closed_prices.append(barrier_call_closed(S0, K, B, r, sig, T))
    mc_price, _ = barrier_call_mc(S0, K, B, r, sig, T, m=252, N=5000, seed=2)
    mc_prices.append(mc_price)

plt.figure(figsize=(6, 4))
plt.plot(sigma_values, closed_prices, marker="o", label="Analytical")
plt.plot(sigma_values, mc_prices, marker="s", label="MC")
plt.xlabel("sigma")
plt.ylabel("Option price")
plt.title("Comparison with respect to sigma")
plt.legend()
plt.grid(True)
plt.show()


# Convergence analysis in m

m_values = [12, 24, 52, 100, 252, 500]
plain_errors = []
corrected_errors = []

for m in m_values:
    C_mc_m, _ = barrier_call_mc(S0, K, B, r, sigma, T, m, N=10000, seed=3)

    B_m = B * np.exp(beta1 * sigma * np.sqrt(T / m))
    C_corr_m = barrier_call_closed(S0, K, B_m, r, sigma, T)

    plain_errors.append(abs(C_mc_m - C_exact))
    corrected_errors.append(abs(C_mc_m - C_corr_m))

plt.figure(figsize=(6, 4))
plt.plot(m_values, plain_errors, marker="o", label="Uncorrected error")
plt.plot(m_values, corrected_errors, marker="s", label="Corrected error")
plt.xlabel("m")
plt.ylabel("Absolute error")
plt.title("Convergence analysis")
plt.legend()
plt.grid(True)
plt.show()
```

The Monte Carlo price is slightly higher than the analytical continuous barrier price, which is expected since discrete monitoring misses barrier crossings.

After applying the correction, the analytical price with shifted barrier is closer to the Monte Carlo estimate, and the absolute error decreases.

From the comparison plot with option price against $\sigma$, both analytical and Monte Carlo prices decrease as $\sigma$ increases, and the Monte Carlo estimates remain close to the analytical values.

From the convergence plot, the uncorrected error is significantly larger and does not decrease consistently, while the corrected error is much smaller and decreases as $m$ increases, consistent with a convergence rate of order $O(1/\sqrt{m})$.

Question 3

For the implicit scheme we solve the Black-Scholes PDE

$$
\partial_t c + \frac{1}{2}\sigma^2 S^2 \partial_{SS} c + rS \partial_S c - rc = 0
$$

on $0 \leq S \leq B$ and $0 \leq t < T$, with terminal condition

$$
c(S,T) = (S-K)^+ 1_{\{S < B\}}
$$

and boundary conditions

$$
c(0,t) = 0
\qquad
c(B,t) = 0
$$

Using the implicit scheme, at each time step we solve a tridiagonal linear system. We then compute the delta 

$$
\Delta(S_i,t_n) \approx \frac{c_{i+1}^n - c_{i-1}^n}{2\Delta S}
$$

```python

# Implicit BTCS 

def barrier_call_implicit(K, B, r, sigma, T, M=200, N=200):
    # grid
    S = np.linspace(0, B, M + 1)
    t = np.linspace(0, T, N + 1)

    dS = S[1] - S[0]
    dt = t[1] - t[0]

    # option grid
    V = np.zeros((N + 1, M + 1))

    # terminal payoff
    V[-1, :] = np.maximum(S - K, 0.0)
    V[-1, -1] = 0.0

    # boundary conditions
    V[:, 0] = 0.0
    V[:, -1] = 0.0

    # build tridiagonal matrix A
    m = M - 1
    A = np.zeros((m, m))

    for i in range(1, M):
        Si = S[i]

        a = 0.5 * sigma**2 * Si**2 / dS**2 - 0.5 * r * Si / dS
        b = -sigma**2 * Si**2 / dS**2 - r
        c = 0.5 * sigma**2 * Si**2 / dS**2 + 0.5 * r * Si / dS

        row = i - 1

        if row > 0:
            A[row, row - 1] = -dt * a
        A[row, row] = 1 - dt * b
        if row < m - 1:
            A[row, row + 1] = -dt * c

    # backward in time
    for n in range(N - 1, -1, -1):
        rhs = V[n + 1, 1:M].copy()
        V[n, 1:M] = np.linalg.solve(A, rhs)

    return S, t, V


# parameters

K = 95
B = 120
r = 0.03
sigma = 0.20
T = 1.0
S0 = 100

# solve PDE
S, t, V = barrier_call_implicit(K, B, r, sigma, T, M=200, N=200)

# price at t=0
price = np.interp(S0, S, V[0, :])
print("Implicit FD price:", price)


# sensitivity analysis in sigma

sigma_vals = [0.10, 0.15, 0.20, 0.25, 0.30]
prices = []

for sig in sigma_vals:
    S_tmp, t_tmp, V_tmp = barrier_call_implicit(K, B, r, sig, T, M=150, N=150)
    prices.append(np.interp(S0, S_tmp, V_tmp[0, :]))

plt.figure(figsize=(6, 4))
plt.plot(sigma_vals, prices, marker="o")
plt.xlabel("sigma")
plt.ylabel("price")
plt.title("Sensitivity with respect to sigma")
plt.grid(True)
plt.show()


# barrier option surface c(S,t)

T_mesh, S_mesh = np.meshgrid(t, S)

fig = plt.figure(figsize=(8, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(S_mesh, T_mesh, V.T, cmap="viridis")
ax.set_xlabel("S")
ax.set_ylabel("t")
ax.set_zlabel("c(S,t)")
ax.set_title("Barrier option surface")
plt.show()


# delta at t=0
dS = S[1] - S[0]
delta = np.zeros_like(S)
delta[1:-1] = (V[0, 2:] - V[0, :-2]) / (2 * dS)

plt.figure(figsize=(6, 4))
plt.plot(S, delta)
plt.xlabel("S")
plt.ylabel("Delta")
plt.title("Delta as a function of S")
plt.grid(True)
plt.show()
```

The implicit finite difference price is close to the analytical value obtained earlier.

From the sensitivity plot with respect to $\sigma$, the option value decreases as $\sigma$ increases. This is expected for an up-and-out call, since higher volatility increases the probability of hitting the barrier, which reduces the option value.

From the surface plot $c(S,t)$, the option value is zero at the barrier $S=B$, which reflects the knock-out condition. For $S$ far below the barrier, the option behaves like a standard call option.

The delta plot shows that for small $S$, the delta is close to zero, since the option is far out-of-the-money. For in between values of $S$, the delta becomes positive, similar to a standard call option. Close to the barrier, the delta decreases and becomes negative, since an increase in $S$ increases the probability of hitting the barrier.
