Question 1.1

We apply Ito's formula to $\log{S_t}$:

$$
df(S_t) = f'(S_t)\, dS_t + \frac{1}{2}f''(S_t)\, d[S_t]
$$

with $f(S_t) = \log{S_t}$ and $dS_t = rS_t \, dt + \sigma_t S_t \, dB_t$. We use the following derivatives
\begin{align}
f'(S_t) &= \frac{1}{S_t} \\
f''(S_t) & = -\frac{1}{S_t^2} \\
d[S_t] & = \sigma_t^2 S_t^2 \, dt
\end{align}

Subsituting this in Ito's formula gives

\begin{align}
d(\log{S_t}) &= r\, dt + \sigma_t \, dB_t - \frac{1}{2} \sigma_t^2 \, dt \\
&= (r- \frac{1}{2} \sigma_t^2) \, dt + \int^{T}_0 \sigma_u \, dB_u
\end{align}

Integrating this expression gives the following

$$
\log{\frac{S_T}{S_0}} = \int^T_0 (r- \frac{1}{2} \sigma_t^2) \, dt + \int^T_0 \sigma_u \, dB_u
$$

We now divide the SDE by $S_t$ and integrate it to put it in a similar form

\begin{align}
\int^T_0 \frac{dS_u}{S_u} &= \int^T_0 r \, dt + \int^T_0 \sigma_u \, dB_u \\
&= rT + \int^T_0 \sigma_u \, dB_u
\end{align}

Now we rewrite the expression in Ito's formula and subsitute the previous equation in to get the following

\begin{align}
\int^T_0 (r - \frac{1}{2} \sigma_u^2) \, du &= \log{\frac{S_T}{S_0}} - \int^T_0 \sigma_u \, dB_u \\
rT - \frac{1}{2} \int^T_0 \sigma_u^2 \, du &= \log{\frac{S_T}{S_0}} - \int^T_0 \sigma_u \, dB_u \\
-2rT + \int^T_0 \sigma_u^2 \, du &= -2 \log{\frac{S_T}{S_0}} + 2 \int^T_0 \sigma_u \, dB_u \\
\int^T_0 \sigma_u^2 \, du &= - 2 \log{\frac{S_T}{S_0}} + 2rT +2 \int^T_0 \sigma_u \, dB_u \\
\int^T_0 \sigma_u^2 \, du &= - 2 \log{\frac{S_T}{S_0}} + 2rT - 2rT + 2 \int^T_0 \frac{dS_u}{S_u}\\
\end{align}

which equates the desired expression

$$
\int^T_0 \sigma_u^2 \, du = 2 \int^T_0 \frac{dS_u}{S_u} - 2 \log{\frac{S_T}{S_0}} 
$$

We rewrite the t-time price process using the expression in question 1.1 to get the following

\begin{align}
\Pi_t &= e^{-r(T-t)}\mathbb{E}_{\mathbb{Q}}[\int^T_0 \sigma_u^2  \, du | \mathcal{F}_t] \\
&= 2e^{-r(T-t)}\mathbb{E}_{\mathbb{Q}}[\int_0^T \frac{dS_u}{S_u} | \mathcal{F}_t] -2e^{-r(T-t)}\mathbb{E}_{\mathbb{Q}}[\log{\frac{S_T}{S_0}} | \mathcal{F}_t] 
\end{align}

We define the second part of the expression to be $L_t$ and for the first part we split the integral

$$
\int_0^T \frac{dS_u}{S_u} = \int_0^t \frac{dS_u}{S_u} + \int^T_t  \frac{dS_u}{S_u}
$$

The first part of this is now $\mathcal{F}_t$-measurable so we can get it out of the expression, we obtain

$$
\mathbb{E}_{\mathbb{Q}}[\int^T_0 \frac{dS_u}{S_u} | \mathcal{F}_t] = \int^t_0 \frac{dS_u}{S_u} + \mathbb{E}_{\mathbb{Q}}[\int^T_t \frac{dS_u}{S_u} | \mathcal{F}_t]
$$

We now substitute the SDE in for the second part and get

\begin{align}
\mathbb{E}_{\mathbb{Q}}[\int^T_t \frac{dS_u}{S_u} | \mathcal{F}_t] &= \mathbb{E}_{\mathbb{Q}}[\int^T_t r \, du | \mathcal{F}_t] + \mathbb{E}_{\mathbb{Q}}[\int^T_t \sigma_u \, dB_u | \mathcal{F}_t] \\
&= r(T-t) + 0
\end{align}

where we used the fact that the expectation of a stochastic integral is equal to 0. Putting it all together we finally obtain

\begin{align}
\Pi_t &= 2e^{-r(T-t)}[\int^t_0 \frac{dS_u}{S_u} + r(T-t)] + L_t \\
&= L_t + 2(T-t)re^{-r(T-t)} + 2e^{-r(T-t)}\int^t_0 \frac{dS_u}{S_u} 
\end{align}

We discount this and substitute this in the formula of $L_t$ and get

$$

where we use the fact that the first term is $\mathcal{F}_t$-measurable and omitted the indicators for simplicity. For the call and put options we get the following

C(t,T,v) &= e^{-r(T-t)}\mathbb{E}_\mathbb{Q}[(S_T-v)_+ | \mathcal{F}_t] \\

\Pi_t = L_t + 2(T-t)re^{-r(T-t)} + 2e^{-r(T-t)}\int^t_0 \frac{dS_u}{S_u} 
$$

with 

$$
L_t := -2e^{-r(T-t)}\log(F/S_0) -\frac{2}{F}e^{-r(T-t)}\mathbb{E}_\mathbb{Q}[S_T-F| \mathcal{F}_t]+\int^\infty_F \frac{2}{v^2}C(t,T,v) \, dv+ \int^F_0 \frac{2}{v^2}P(t,T,v) \, dv
$$ 

The VIX estimator is given as 


$$
VIX_t := \sqrt{\frac{2e^{r\tau}}{\tau} ( \int_0^{F_{t,t+\tau}} \frac{1}{K^2}P(t,t+\tau,K)\, dK + \int_{F_{t,t+\tau}}^\infty \frac{1}{K^2}C(t,t+\tau,K)\, dK )}
$$

The connection to the VIX estimator is that it essentially uses the same put and call options part in the formula for $L_t$. In particular, both formulas contain the integrals of put options and call options, weighted by $\frac{1}{K^2}$. The VIX estimator then rescales this quantity by $\frac{2e^{r\tau}}{\tau}$ and takes the square root to obtain the annualized market-implied volatility.

The advantages of this estimator are that it is forward looking since its based on current and not past option prices and thus can be interpreted as an average of future volatility. Moreover it does not rely on a stochastic model so the interpretation of the estimator is very direct.

The disadvantages are that it assumes a continuum of option prices while we might only have a finite number available. This means that we might have to interpolate data which introduces additional errors.

Constructing the variance estimator from logarithmic returns instead of arithmetic returns gives us the advantage that the log-returns are additive over time and can thus be summed up to obtain the total return. However, the log-returns are not as intuitive as arithmetic returns since price changes cannot be interpreted directly as percentage changes.


Question 1.5

To check that the portfolio $V_t$ is self-financing we take the derivative with respect to t. If all changes in $V_t$ are due only to the price movements of the underlying assets and accumulation of interests then it is self-financing. Essentially, it needs to satisfy

$$
dV_t = dL_t + \phi \, dS_t + \psi \, dA_t
$$

We have that 

$$
V_t = L_t + 2(T-t)re^{-r(T-t)} + 2e^{-r(T-t)}\int^t_0 \frac{dS_u}{S_u} 
$$

with the derivatives 

\begin{align}
d(2(T-t)re^{-r(T-t)}) &= 2re^{-r(T-t)}((T-t)r-1)\, dt \\
d(2e^{-r(T-t)}\int^t_0 \frac{dS_u}{S_u}) &= 2re^{-r(T-t)}(\int^t_0 \frac{dS_u}{S_u}) \, dt + 2e^{-r(T-t)} \frac{dS_t}{S_t}
\end{align}

Putting it together we obtain

$$
dV_t = dL_t + 2re^{-r(T-t)}((T-t)r-1)\, dt + 2re^{-r(T-t)}(\int^t_0 \frac{dS_u}{S_u}) \, dt + 2e^{-r(T-t)} \frac{dS_t}{S_t}
$$

We use the fact that $\phi S_t = 2e^{-r(T-t)}$ and so $\phi \, dS_t = 2e^{-r(T-t)} \frac{dS_t}{S_t}$. Similarly $A_t = e^{rt}$, so $dA_t = rA_t$. Using this we get

$$
\psi \, dA_t = 2e^{-rT}(\int^t_0 \frac{dS_u}{S_u}+(T-t)r-1)re^{rt} = 2re^{-r(T-t)}((T-t)r-1)\, dt + 2re^{-r(T-t)}(\int^t_0 \frac{dS_u}{S_u}) \, dt
$$

Substituting these in $dV_t$ gives

$$
dV_t = dL_t + \phi \, dS_t + \psi \, dA_t
$$

which is exactly the self-financing property.

Question 1.6

We set $u(t):= \mathbb{E}_{\mathbb{Q}}[V_t]$ and compute using Fubini

$$
\frac{1}{T} \mathbb{E}_{\mathbb{Q}}[\int^T_0 V_t \, dt] = \frac{1}{T} \int^T_0 u(t) \, dt
$$

We now take expectations under measure $Q$ of the Heston variance price process

\begin{align}
\mathbb{E}_{\mathbb{Q}}[dV_t] &= \kappa \mathbb{E}_{\mathbb{Q}}[\theta - V_t] \, dt + \sigma \mathbb{E}_{\mathbb{Q}}[\sqrt{V_t} /, dW_t] \\
&= \kappa (\theta - \mathbb{E}_{\mathbb{Q}}[V_t]) \, dt \\
du(t) &= \kappa (\theta - u(t)) \, dt
\end{align}

where we used the fact that the mean of a stochastic integral is 0 and substituted in the definition of $u(t)$.
This gives us the following ODE

$$
\frac{du(t)}{dt} = \kappa (\theta - u(t)), \quad u(0)= v_0
$$

with solution 

$$
u(t) = \theta + (v_0 - \theta)e^{- \kappa t}
$$

Integrating this over $[0, T]$ gives

\begin{align}
\frac{1}{T} \int^T_0 u(t) \, dt &= \frac{1}{T} \int^T_0 \theta + (v_0 - \theta)e^{- \kappa t} \, dt \\
&= \frac{1}{T} \int^T_0 \theta \, dt + (v_0 - \theta) \frac{1}{T} \int^T_0 e^{- \kappa t} \, dt \\
&= \theta + \frac{v_0 - \theta}{\kappa T}(1-e^{-\kappa T})
\end{align}

Thus

$$
\frac{1}{T} \mathbb{E}_{\mathbb{Q}}[\int^T_0 V_t \, dt] = \theta + \frac{v_0 - \theta}{\kappa T}(1-e^{-\kappa T})
$$
